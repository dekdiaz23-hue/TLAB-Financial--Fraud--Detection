# Data Cleaning & Preprocessing

Hypothesis: Fraud is concentrated in TRANSFER and CASH_OUT transactions and origin account has a balance thatm is fully drained to 0

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
df=pd.read_csv('../data/Tlab.csv')

```

## 1.Drop irrelevant columns: nameOrig, isFlaggedFraud and nameDest

```python

df.drop(columns=['nameOrig', 'nameDest', 'isFlaggedFraud'], inplace=True)
```

## 2. Filter to TRANSFER and CASH_OUT only since fraud only exists there

```python

df = df[df['type'].isin(['TRANSFER', 'CASH_OUT'])]
```

## 3. Create a feature for account balance drained to 0 and also inlcudinding large erros that would be make a transaction more likely to be fraudulent since that would be suspicious

```python
df['isAccountDrained'] = (df['newbalanceOrig'] == 0).astype(int)
df['errorBalanceOrig'] = df['newbalanceOrig'] + df['amount'] - df['oldbalanceOrg']
```

```python
# One-hot encode the 'type' column
df = pd.get_dummies(df, columns=['type'], drop_first=True)
print(f'Shape after encoding: {df.shape}')
df.head()
```
