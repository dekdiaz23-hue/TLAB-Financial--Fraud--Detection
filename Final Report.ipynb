# Which insights did you gain from your EDA?
Fraud only occurs in TRANSFER and CASH_OUT transactions — every other transaction type has zero fraud cases. 

The dataset is severely imbalanced with only 0.13% of transactions being fraudulent. 

Fraudulent transactions almost always drain the origin account completely to zero after the transaction. Fraud transactions also average about 8x higher in amount compared to legitimate ones. 

The existing isFlaggedFraud column catches only 16 out of 8,213 fraud cases. It was stated that it was a naive model but EDA showed how bad it actually is. 


# How did you determine which columns to drop or keep?
isFlaggedFraud was dropped because its the old naive model. 
The dataset was filtered to only TRANSFER and CASH_OUT because EDA proved fraud never occurs in any other transaction type
The other columns that held id information was dropped because it was irrelevant to tryign to find fraud. 


# Which hyperparameter tuning strategy did you use? Why?
RandomizedSearchCV. The data set was more than 2 million and gradientboosting would have taken longer and I would not be able to acccount for the class imbalance as easy.
